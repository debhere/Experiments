```python
import pymupdf
import pandas as pd
```

```python
sf = "lic’s_bima_kavach_cis.pdf"
```

```python
dfs = []
pages = pymupdf.open(sf)

for page in pages:
    tables = page.find_tables()
    for table in tables:
        df = table.to_pandas()
        dfs.append(df)
```
