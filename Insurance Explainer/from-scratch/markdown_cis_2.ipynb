```python
import pymupdf
import pandas as pd
```

```python
sf = "jeevan_utsav_single_premium_cis.pdf"
```

```python
dfs = []
pages = pymupdf.open(sf)

for page in pages:
    tables = page.find_tables()
    for table in tables:
        df = table.to_pandas()
        dfs.append(df)
```

```python
def claim_func(df, idx):

    claim = df.loc[idx:].reset_index()
    field_map = {}
    for idx, row in claim.iterrows():
        if idx == 0:
            fields = list(row.values[3:7])
            desc = ["", "", "", "", "", ""]
            field_map = dict(zip(fields, desc))
            continue
    
        fields = list(field_map.keys())
        
        field_map[fields[0]] = field_map[fields[0]] + "" if pd.isna(row.values[3]) else row.values[3]
        field_map[fields[1]] = field_map[fields[1]] + "" if pd.isna(row.values[4]) else row.values[4]
        field_map[fields[2]] = field_map[fields[0]] + "" if pd.isna(row.values[5]) else row.values[5]
        field_map[fields[3]] = field_map[fields[0]] + "" if pd.isna(row.values[6]) else row.values[6]
    
    # print(field_map)
    return field_map
```

```python
def table_cont(df0, df1):
        header = ""
        if df1.columns[1] == 'Col1':
            if not pd.isna( df0.tail(1).iloc[:, 1].values[0]):
                header += df0.tail(1).iloc[:, 1].values[0].replace('\n', ' ').replace('/', 'and')
        txt = f"## {header} \n\n### Description\n{df0.tail(1).iloc[:, 2].values[0]}{df1.columns[2]}\n\n### Policy Clause\n{df0.tail(1).iloc[:, 3].values[0]}"
        return txt
    # print(txt)
```

```python
dfs[0]
```

```python
contents = ""

for i, df in enumerate(dfs):
    for idx, row in df.iterrows():
        cols = df.columns
        if cols[1] == 'Col1':
            contents += table_cont(dfs[i-1], dfs[i])

        if pd.isna(row.values[1]):
            continue
        header = row.values[1].replace('\n', ' ').replace('/', 'and')
        description = row.values[2]
        policy_clause = df.tail(1).iloc[:, len(cols) - 1].values[0]
        
        if 'claim' in row.values[1].lower() or 'grievances' in row.values[1].lower() or 'servicing' in row.values[1].lower():
            desc_fields = claim_func(df, idx + 1)
            description += "\n"
            for f, d in desc_fields.items():
                description += f"{f}: {d}\n"
        t = f"## {header}\n\n ### Description\n{description}\n\n### Policy Clause\n{policy_clause}\n\n"
        contents += '\n\n'
        contents += t
        print(contents)
```
